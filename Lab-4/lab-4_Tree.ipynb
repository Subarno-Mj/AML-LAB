```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv("D:\Practice_SMJ\Social_Network_Ads.csv")
```

```python
display(df)
```

```python
df.head()
```

```python
# x = age 
# y = estimated salary 
# we will predict purchaed or not 
df.isnull().sum() 
```

```python
X = df[["Age", "Estimated_Salary"]]  # input / feature
y = df["Purchased"]   # target / output
```

```python
print(X)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

```

```python
sc = StandardScaler()
X_train= sc.fit_transform(X_train)
X_test = sc.transform(X_test)
X_test
```

```python
# fitting decision tree classification to traning set
classifier = DecisionTreeClassifier(criterion="entropy", random_state=0)
classifier.fit(X_train, y_train)
```

```python
#predicting test set results
y_pred = classifier.predict(X_test)

print("Predicted values:-")
print(y_pred)

print("Actual values:-")
print(y_test.values)
```

```python
y_pred
```

```python
y_test.values
```

```python
from sklearn.metrics import confusion_matrix
# maring confusion Matrix to compare
cm = confusion_matrix(y_test , y_pred)
cm
```

```python
accuracy = (cm[0][0]+cm[1][1])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1] )
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy * 100, "%")
print("Error:", (1 - accuracy) * 100, "%")

```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
# predict for a new person with age = 29 , salary = 80000
new_person = np.array([[29,80000]])

# scale new person features
new_person_scale= sc.transform(new_person)

#predict using trained classifier
prediction = classifier.predict(new_person_scale)

if prediction[0]==1:
    print("The individual is predicted to purchase..")
else:
    print("The individual is predicted not to purchase...")
```
