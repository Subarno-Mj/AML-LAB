```python
# 3------

# 1. Basic Lambda Function (Single Argument)
# Squares a given number
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")  # Output: 25

# 2. Lambda with Multiple Arguments
# Multiplies two numbers together
multiply = lambda a, b: a * b
print(f"Product of 4 and 6: {multiply(4, 6)}")  # Output: 24

# 3. Lambda with Conditional Logic (if-else)
# Checks if a number is even or odd
check_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(f"The number 7 is: {check_even(7)}")  # Output: Odd

# 4. Using Lambda inside map()
# Cubes every number in the list
numbers = [1, 2, 3, 4]
cubed_numbers = list(map(lambda n: n ** 3, numbers))
print(f"Cubed list: {cubed_numbers}")  # Output: [1, 8, 27, 64]

# 5. Using Lambda inside filter()
# Extracts only the even numbers from the list
mixed_numbers = [12, 5, 8, 19, 22, 3]
even_numbers = list(filter(lambda x: x % 2 == 0, mixed_numbers))
print(f"Filtered even numbers: {even_numbers}")  # Output: [12, 8, 22]

# 6. Using Lambda inside sorted()
# Sorts a list of tuples based on the second value (the age)
people = [("Alice", 25), ("Bob", 20), ("Charlie", 30)]
sorted_by_age = sorted(people, key=lambda person: person[1])
print(f"Sorted by age: {sorted_by_age}")  
# Output: [('Bob', 20), ('Alice', 25), ('Charlie', 30)]
```

```python
#5--------Quicksort using Recursion

def quicksort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]  # Choose middle element as pivot

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quicksort(left) + middle + quicksort(right)


numbers = [3, 6, 8, 10, 1, 2,1]

sorted_numbers = quicksort(numbers)

print("Original list:", numbers)
print("Sorted list:", sorted_numbers)
```
