```python
# for independent and dependent variable do column split

#split independent varibales
x=df.iloc[:,:-1].values      # show data for all column except the last column i.e -1
print(x)
```

```python
from sklearn.impute import SimpleImputer

# To handle missing values
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

# Calculate the mean and fit the imputer
imputer = imputer.fit(x[:, 1:3])

# Replace missing values with the mean
x[:, 1:3] = imputer.transform(x[:, 1:3])

# Print the updated data
print(x)
```

```python
# Encoding categorical data
# Encoding independent variable

from sklearn.preprocessing import LabelEncoder, OneHotEncoder

labelEncoder_x = LabelEncoder()

x[:, 0] = labelEncoder_x.fit_transform(x[:, 0])

print(x)
```

```python
# Encoding the inedependent variable 
from sklearn.compose import ColumnTransformer

ct = ColumnTransformer([("Country", OneHotEncoder(),[0])],remainder="passthrough")
x=ct.fit_transform(x)
x.astype(int)
```

```python
#Encoding the depoendent variable
from sklearn.preprocessing import LabelEncoder
labelEncoder_y = LabelEncoder()
y= labelEncoder_y.fit_transform(y)
y
```

```python
# Split data into Training set and Test set
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=45)

```
