```python
#10. Add Grade Column
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(grade)

print(df)
```

```python
#12. Missing Values
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

print("DataFrame:")
print(students)

print("\nMissing values:")
print(students.isnull())

print("\nNumber of missing values in each column:")
print(students.isnull().sum())

print("\nRows containing missing values:")
print(students[students.isnull().any(axis=1)])
```

```python
#13. Replace Missing Values with Average
students["Age"] = students["Age"].fillna(students["Age"].mean())

students["Marks"] = students["Marks"].fillna(students["Marks"].mean())

print(students)
```

```python
#14. Remove Rows Containing Missing Values

students = students.dropna() # updates the actual dataframe

print(students ,"\n")

#don't want to permanently change the DataFrame then
print(students.dropna())
```

```python
#15. Find and Remove Duplicates

data = {
    "Name": ["Asha", "Ravi", "Meena", "Asha", "Arun"],
    "Marks": [85, 78, 92, 85, 88]
}

df2 = pd.DataFrame(data)

print("DataFrame:")
print(df2)

print("\nDuplicate rows:")
print(df2[df2.duplicated()])

print("\nAfter removing duplicates:")
print(df2.drop_duplicates())
```

```python
#16. Statistical Calculations
marks = pd.DataFrame(
    {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
    }
)

print("Mean:", marks["Marks"].mean())
print("Median:", marks["Marks"].median())
print("Maximum:", marks["Marks"].max())
print("Minimum:", marks["Marks"].min())
print("Standard deviation:", marks["Marks"].std())
print("Sum:", marks["Marks"].sum())
```

```python
#17. GroupBy — Department Analysis
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],
    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],
    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)
```

```python
#Average marks by department

print(df.groupby("Department")["Marks"].mean())
```

```python
#Maximum marks by department

print(df.groupby("Department")["Marks"].max())
```

```python
#Number of students

print(df.groupby("Department")["Name"].count())
```

```python
df.groupby("Department")["Marks"].agg(["mean", "min", "max", "count"])
```

```python
# 2nd way
print(
    df.groupby("Department")["Marks"].agg
    (
        ["mean", "min", "max", "count"]
    )
)
```

```python
#18. Combine Two DataFrames Using StudentID

students = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Meena", "Kiran"]
})

marks = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "Marks": [85, 78, 92, 74]
})

result = pd.merge(students, marks, on="StudentID")

print(result)
```

```python
#19. Student Performance Analysis

#Create DataFrame
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
#First five records
print(df.head())
```

```python
#Dimensions
print(df.shape)
```

```python
#Average final mark
print(df["FinalMarks"].mean())
```

```python
#Highest-scoring student
print(df.loc[df["FinalMarks"].idxmax()])
```

```python
#Lowest-scoring student
print(df.loc[df["FinalMarks"].idxmin()])
```

```python
#Students scoring above 80
print(df[df["FinalMarks"] > 80])
```

```python
#Average marks by department
print(df.groupby("Department")["FinalMarks"].mean())
```

```python
#Students having attendance above 80%
print(df[df["Attendance"] > 80])
```

```python
#Sort by FinalMarks

print(df.sort_values("FinalMarks", ascending=False))
```

```python
#Add Result column
df["Result"] = np.where(
    df["FinalMarks"] >= 70,
    "Pass",
    "Fail"
)

print(df)
```

```python
#Add Performance Category
def performance(marks):
    if marks >= 90:
        return "Excellent"
    elif marks >= 80:
        return "Very Good"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["FinalMarks"].apply(performance)

print(df)
```

```python
#Department with highest average FinalMarks
avg_department = df.groupby("Department")["FinalMarks"].mean()

print(avg_department)

print("\nDepartment with highest average:")
print(avg_department.idxmax())
```

```python
#20. Employee Analysis

#create dataframe

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

df
```

```python
# Average salary
print(employees["Salary"].mean())
print("\n")
#Maximum salary
print(employees["Salary"].max())
print("\n")
#Minimum salary
print(employees["Salary"].min())
print("\n")
#Employee receiving highest salary
print(employees.loc[employees["Salary"].idxmax()])
```

```python
#Employees earning above ₹60,000

print(employees[employees["Salary"] > 60000])
```

```python
#Average experience by department

print(employees.groupby("Department")["Experience"].mean())
```

```python
#Number of employees in each department
print(employees.groupby("Department")["Employee"].count())
```

```python
#IT employees with more than 3 years experience
print(
    employees[
        (employees["Department"] == "IT") &
        (employees["Experience"] > 3)
    ]
)
```

```python
#Add SalaryLevel column
def salary_level(salary):
    if salary >= 75000:
        return "High"
    elif salary >= 50000:
        return "Medium"
    else:
        return "Low"

employees["SalaryLevel"] = employees["Salary"].apply(salary_level)

print(employees)
```
