# **9. Boolean Indexing**

```python
import numpy as np

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

# Marks greater than 75
print("Marks > 75:", marks[marks > 75])

# Marks between 50 and 80 (inclusive)
print("Marks between 50 and 80:", marks[(marks >= 50) & (marks <= 80)])

# Number of students scoring below 50
print("Number of students scoring below 50:", np.sum(marks < 50))

# Percentage of students who scored 60 or above
pct_60_or_above = (np.sum(marks >= 60) / len(marks)) * 100
print(f"Percentage scoring >= 60: {pct_60_or_above:.1f}%")
```

# **10.  2D-Array Marks Analysis**

```python
import numpy as np

marks = np.array([
                  [80, 75, 90],
                   [65, 70, 72],
                   [88, 92, 85],
                    [55, 60, 58]
                 ])

tot_per_student = np.sum(marks, axis=1)
avg_per_student = np.mean(marks, axis=1)
avg_per_subject = np.mean(marks, axis=0)
highest_student = np.argmax(tot_per_student) + 1
highest_per_subject = np.max(marks, axis=0)

print("Total marks of each student:", tot_per_student)
print("Average marks of each student:", avg_per_student)
print("Average marks in each subject:", avg_per_subject)
print(f"Student with highest total: Student {highest_student} ({np.max(tot_per_student)} marks)")
print("Highest mark in each subject:", highest_per_subject)
```

# **12. Daily Temperature Analysis**

```python
import numpy as np

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

mean_temp = np.mean(temperature)
max_temp = np.max(temperature)
min_temp = np.min(temperature)

print(f"Mean temperature: {mean_temp:.2f}°C")
print(f"Max temperature: {max_temp}°C | Min temperature: {min_temp}°C")
print("Number of days above average temp:", np.sum(temperature > mean_temp))
print("Temperatures > 34°C:", temperature[temperature > 34])
print("Difference (Max - Min):", max_temp - min_temp)
print(f"Standard deviation: {np.std(temperature):.2f}")
```

# **13. Monthly Sales Determination**

```python
import numpy as np

sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Total annual sales: ₹", np.sum(sales))
print("Average monthly sales: ₹", np.mean(sales))
print("Month with highest sales: Month", np.argmax(sales) + 1)
print("Month with lowest sales: Month", np.argmin(sales) + 1)
print("Months with sales > ₹20,000:", np.sum(sales > 20000))

# Percentage increase from Jan (index 0) to Dec (index -1)
pct_increase = ((sales[-1] - sales[0]) / sales[0]) * 100
print(f"Percentage increase Jan to Dec: {pct_increase:.2f}%")
```

# **14. Student Performance Categorization (np.select)**

```python
import numpy as np

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

tot_marks = np.sum(marks, axis=1)
avg_marks = np.mean(marks, axis=1)
subject_avg = np.mean(marks, axis=0)

print("Total marks per student:", tot_marks)
print("Average per student:", avg_marks)
print("Average mark per subject:", subject_avg)
print("Highest-scoring student: Student", np.argmax(tot_marks) + 1)
print("Lowest-scoring student: Student", np.argmin(tot_marks) + 1)
print("Students with average > 80: Student(s)", np.where(avg_marks > 80)[0] + 1)
print(f"Overall class average: {np.mean(marks):.2f}")

# Assigning performance categories using np.select()
conditions = [
    avg_marks >= 85,
    (avg_marks >= 70) & (avg_marks < 85),
    (avg_marks >= 50) & (avg_marks < 70)
]
choices = ['Excellent', 'Good', 'Average']
performance = np.select(conditions, choices, default='Needs Improvement')

for i, category in enumerate(performance):
    print(f"Student {i+1} Category: {category}")
```

# **21. Array Normalization (Min-Max Scaling)**

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50], dtype=float)

# Min-Max Normalization formula: (X - min) / (max - min)
normalized_data = (data - np.min(data)) / (np.max(data) - np.min(data))

print("Original Data:", data)
print("Normalized Data:", normalized_data)
```

# **22. Array Standardization (Z-Score)**

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50], dtype=float)

# Z-Score Standardization formula: (X - mean) / std
standardized_data = (data - np.mean(data)) / np.std(data)

print("Original Data:", data)
print("Standardized Data:", standardized_data)
```

# **24. Performance Comparison (Python List vs. NumPy Array)**

```python
import numpy as np
import time

N = 1_000_000

# Python List Multiplication
list1 = list(range(1, N + 1))
list2 = list(range(1, N + 1))

t0 = time.time()
list_result = [x * y for x, y in zip(list1, list2)]
t1 = time.time()
py_list_time = t1 - t0

# NumPy Array Multiplication
arr1 = np.arange(1, N + 1, dtype=np.int64)
arr2 = np.arange(1, N + 1, dtype=np.int64)

t0 = time.time()
numpy_result = arr1 * arr2
t1 = time.time()
numpy_time = t1 - t0

print(f"Python List Time : {py_list_time:.5f} seconds")
print(f"NumPy Array Time : {numpy_time:.5f} seconds")
print(f"Speedup Factor   : NumPy is {py_list_time / numpy_time:.2f}x faster")
```
