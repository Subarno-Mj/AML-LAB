```python
# Create a 1D array
arr = np.array([1, 2, 3, 4, 5, 6])

# Reshape into 2 rows and 3 columns
reshaped_arr = arr.reshape(2, 3)

print("Original:", arr)
print("Reshaped (2x3):\n", reshaped_arr)
```

```python
# create 3x3 matrix
n = np.matrix([[1,2,3],
               [5,6,7],
               [55,66,44]]

)
print(n)
type(n)
print("Main Diagonal = ",np.diag(n))
print("Elements above main Diagonal = ", np.diag(n,1))
print("Elements below main Diagonal = ", np.diag(n,-1))
print("Elements one step below main Diagonal = ", np.diag(n,-2))
```

```python
# Repeating list
arr = np.array([1,2,3]*3)
print(arr)

# Repeating elements
arr = np.repeat([1,2,3],3)
print(arr)
```

```python
# Vertical Stack Row wise
p= np.ones([3,3],int)
print(p)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Stack vertically (creates a 2D array)
v_stacked = np.vstack((a, b))

print(v_stacked)

# Hstack
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Stack horizontally
h_stacked_1d = np.hstack((a, b))

print(h_stacked_1d)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Doing Add and Sub element wise in array
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**2) # power square
print(a@b)  # dot product 1*4 + 2*5 + 3*6 =32
```

```python
# Notice the outer brackets around all three rows
z = np.array([[4, 5, 6],
              [4, 5, 6],
              [4, 5, 6]])

print(len(z))

#shape
print(z.shape)

#transpose
print(z.T)
```

```python
# covert int to float type
z = z.astype('f')
print(type(z))
z.dtype
```

```python
# MATH FUNCTIONS
a=np.array([4,0,2,1,2,3])
print(a.sum())
print(a.min())
print(a.max())
print(round(a.std(),3)) # standard deviation
print(a.argmax())   # index location of max element
print(a.argmin())   # index location of min element
```

```python
import numpy as np

# Correct function is 'arange' (array range)
s = np.arange(100)

print(s)
```

```python
# indexing / slicing

s= np.arange(6)**2  # square 1st 6 elements of s
print(s)
print(s[-2]) #2nd last element
```

```python
# print in given range
print(s)
print(s[1:4])  # print from index 1 to 3

s = np.arange(20)

# Syntax: s[start : stop : step]
print(s[0:len(s):2]) # skip 1
```

```python
# reshape

a = np.array([1, 2, 3, 4, 5, 6])

# Reshape into 2x3
b = a.reshape(2, 3)

print("Reshaped b:\n", b)
print("Original a is unchanged:\n", a)

# resize
a = np.array([1, 2, 3, 4, 5, 6])

# Expand to size 8 (2x4) in-place
a.resize(2, 4)

print("Resized a (in-place with zero padding):\n", a)
```
